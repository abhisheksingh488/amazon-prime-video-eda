The insights obtained from this analysis can help Amazon Prime Video improve its business strategy. ## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Remove Duplicates Values
titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)
````

````python
# Handling Missing Values
titles['age_certification'].fillna('Unknown', inplace=True)

titles['genres'].fillna('Unknown', inplace=True)

titles['production_countries'].fillna('Unknown', inplace=True)

titles['imdb_score'].fillna(titles['imdb_score'].mean(), inplace=True)
````

````python
# Convert Data Type
titles['release_year'] = titles['release_year'].astype(int)
````

````python
# Merge Datasets
df = pd.merge(titles, credits, on='id')
````

````python
df.head()
````

* Data types of some variables were checked and corrected where necessary.

* The titles and credits datasets were merged using the id column to combine content information with cast and director details.

### Chart-1-Pie Chart on Show Type (Movies vs TV Shows)

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
plt.figure(figsize=(8,6))

correlation = titles[['release_year','runtime','imdb_score','imdb_votes','tmdb_popularity','tmdb_score']].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap of Numerical Variables")

plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap was chosen because it helps visualize the relationship between multiple numerical variables at once. It makes it easy to identify whether variables are positively or negatively correlated.

##### 2. What is/are the insight(s) found from the chart?

The heatmap shows that some variables, such as IMDb score and TMDB score, have a positive correlation, indicating that titles with higher IMDb ratings tend to have higher TMDB ratings as well.
Other variables show weak or little correlation, suggesting that factors like runtime or release year do not strongly influence the ratings.
